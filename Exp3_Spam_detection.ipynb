```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

df_spam = pd.read_csv('spam.csv')
df_spam['label'] = df_spam['label'].map({'ham': 0, 'spam': 1})

X_train, X_test, y_train, y_test = train_test_split(df_spam['message'], df_spam['label'], test_size=0.2, random_state=42)

vec = TfidfVectorizer(stop_words='english')
X_train_vec = vec.fit_transform(X_train)
X_test_vec = vec.transform(X_test)

model = MultinomialNB()
model.fit(X_train_vec, y_train)

predictions = model.predict(X_test_vec)
print(f"Accuracy: {accuracy_score(y_test, predictions):.4f}")
print("\nClassification Report:\n", classification_report(y_test, predictions))
```

```python
custom_messages = [
    "Hey Abhay, are we still meeting at the campus library tonight?",
    "CONGRATULATIONS! You have been selected to receive a $1000 gift card. Click here to CLAIM now!"
]
custom_vec = vec.transform(custom_messages)
results = model.predict(custom_vec)
for msg, pred in zip(custom_messages, results):
    label = "SPAM" if pred == 1 else "HAM (Normal)"
    print(f"Prediction: {label} | Message: '{msg}'\n")
```
